```python
import pandas as pd
import numpy as np
print(pd.__version__)
```

```python
scores = pd.Series([63, 34, 78])
print(scores)
```

```python
scores = pd.Series([63, 34, 78], index=["Mona", "Omar", "Sara"])
print(scores)
print(scores["Omar"])
```

```python
print(scores + 5)
print(scores.mean())
print(scores.to_numpy())
```

```python
print(scores[scores > 50])
```

```python
quiz = pd.Series({"Sara": 9, "Mona": 7, "Ali": 10})
print(scores + quiz)
```

```python
df = pd.DataFrame({
    "name": ["Mona", "Omar", "Sara", "Ali"],
    "dept": ["CS", "Business", "CS", "Engineering"],
    "hours": [17.2, 1.6, 12.5, 8.0],
    "exam": [63, 34, 78, 55],
})
print(df)
```

```python
print(df.shape)
print(list(df.columns))
print(df.index)
print(df.dtypes)
```

```python
print(type(df["exam"]))
print(df["exam"])
```

```python
print(df[["name", "exam"]])
```

```python
df["points_per_hour"] = (df["exam"] / df["hours"]).round(1)
print(df)
```

```python
print(df.sort_values("exam", ascending=False))
```

```python
print(df.loc[2, "name"])
print(df.iloc[0])
```
